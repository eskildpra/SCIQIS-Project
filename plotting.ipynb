```python
import numpy as np
import matplotlib.pyplot as plt
data = np.load('ladder_ansatz_data.npy')
```

```python
# Increase default text size
plt.rcParams.update({'font.size': 13})

# Assuming 'data' is already defined
n_qubits = [2, 3]
n_layers = [1, 2, 3]
thetas = np.linspace(0, 2*np.pi, 180, endpoint=False)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# Progressive shades for each subplot
colors_2q = [plt.cm.Blues(0.4), plt.cm.Blues(0.7), plt.cm.Blues(1.0)]
colors_3q = [plt.cm.Greens(0.4), plt.cm.Greens(0.7), plt.cm.Greens(1.0)]

# Plot for 2 Qubits
ax1.plot(thetas, data[0, 0, :], label='1 layer', color=colors_2q[0])
ax1.plot(thetas, data[0, 1, :], label='2 layers', color=colors_2q[1])
ax1.plot(thetas, data[0, 2, :], label='3 layers', color=colors_2q[2])

# Plot for 3 Qubits
ax2.plot(thetas, data[1, 0, :], label='1 layer', color=colors_3q[0])
ax2.plot(thetas, data[1, 1, :], label='2 layers', color=colors_3q[1])
ax2.plot(thetas, data[1, 2, :], label='3 layers', color=colors_3q[2])

# Titles placed on the top right
ax1.set_title('2 Qubits', loc='right')
ax2.set_title('3 Qubits', loc='right')

# Compact Legend parameters (Option 1)
legend_kwargs = dict(
    loc='center left',
    bbox_to_anchor=(1.01, 0.5),
    frameon=False,
    handlelength=1.0,           
    handletextpad=0.4,          
    labelspacing=0.2            
)

ax1.legend(**legend_kwargs)
ax2.legend(**legend_kwargs)

# Add transparent gray horizontal line at y=0 on both axes
ax1.axhline(y=0, color='gray', alpha=0.5)
ax2.axhline(y=0, color='gray', alpha=0.5)

# Format the x-axis to be in levels of pi
pi_ticks = [0, np.pi/2, np.pi, 1.5*np.pi, 2*np.pi]
pi_labels = ['0', r'$\pi/2$', r'$\pi$', r'$3\pi/2$', r'$2\pi$']

ax2.set_xticks(pi_ticks)
ax2.set_xticklabels(pi_labels)
ax2.set_xlabel(r'$\theta$', fontsize=17)

# Snap the borders flush with the data
ax1.set_xlim(0, 2*np.pi)

# Shared y-axis label
fig.supylabel('Expected Value', fontsize=17)

# Increase tick size
ax1.tick_params(axis='both', which='major', length=6, width=1.5)
ax2.tick_params(axis='both', which='major', length=6, width=1.5)

# Reclaim plot area (changed right bound from 0.86 to 0.93)
fig.tight_layout(rect=[0, 0, 0.93, 1])

plt.show()
```
