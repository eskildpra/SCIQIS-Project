# Continuous Variable Quantum Information

In continuous variable quantum information information is processed by observables like the srength of a magnetic field. As a classic example one could imagine a box connected to a sping. This is a classic harmonic oscillator where the energy is continuous. At any given time one can determine the momentum and position of the box. For quantum paricles however, this is not true. For quantum particles energy comes in descrete levels instead. One can describe these levels in a quantum harmonic oscillator with Fock states.
$$
|n\rangle
$$
n denotes the level. The energy can be extracted with a hamiltonian given by:
$$
\hat{H} = \hbar \omega (\hat{n}+1/2)
$$
Thus the ground level $|0\rangle$ has an energy of $\hbar \omega \cdot 1/2$

Fock can be mnipulated with creation and annihalation operators:
$$
\hat{a}^{\dagger}|n\rangle = \sqrt{n+1}|n+1\rangle \qquad \hat{a}|n\rangle = \sqrt{n}|n-1\rangle
$$
Measuring observables like the momentum $\hat{P}$ and position $\hat{X}$ can then be done with the creation and annihalation operators like so:
$$
\hat{X} = \cfrac{\hat{a}^\dagger + \hat{a}}{\sqrt{2}} \qquad \hat{P} = \cfrac{\hat{a}^\dagger - \hat{a}}{\sqrt{2}} \cdot i
$$
