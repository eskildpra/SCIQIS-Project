# Continuous Variable Quantum Information

In continuous variable quantum information information is processed by observables like the srength of a magnetic field. As a classic example one could imagine a box connected to a sping. This is a classic harmonic oscillator where the energy is continuous. At any given time one can determine the momentum and position of the box. For quantum paricles however, this is not true. For quantum particles energy comes in descrete levels instead. One can describe these levels in a quantum harmonic oscillator (QHO) with Fock states.
$$
|n\rangle
$$
n denotes the level. The energy can be extracted with a hamiltonian given by:
$$
\hat{H} = \hbar \omega (\hat{n}+1/2)
$$
Thus the ground level $|0\rangle$ has an energy of $\hbar \omega \cdot 1/2$

Fock can be mnipulated with creation and annihalation operators:
$$
\hat{a}^{\dagger}|n\rangle = \sqrt{n+1}|n+1\rangle \qquad \hat{a}|n\rangle = \sqrt{n}|n-1\rangle
$$
Measuring observables like the momentum $\hat{P}$ and position $\hat{X}$ can then be done with the creation and annihalation operators like so:
$$
\hat{X} = \cfrac{\hat{a}^\dagger + \hat{a}}{\sqrt{2}} \qquad \hat{P} = \cfrac{\hat{a}^\dagger - \hat{a}}{\sqrt{2}} \cdot i
$$
The general state of a QHO is given by:
$$
|\psi \rangle = \sum_n c_n |n\rangle \qquad \sum_n |c_n|^2 = 1
$$
When the momentum and position operators are given like this, they no longer commute like they would do classicly.
$$
[\hat{X},\hat{P}] = \cfrac{i}{\hbar}
$$
This leads to the product of the operators uncertainty having a lower bound:
$$
\Delta X \Delta P \geq 1/2
$$
This is called Heisenbergs uncertainty principle. This means that one can never know the exact position and momentum of a quantum paticle at the same time. They will always be an uncertainty on measuring these observables that don't stem from experimental equipment. If one perfectly knows the location of a quantum particle, the momentum must be completely unknown. This can be visualized in a phasespace plot.

Fock states are also called Gassian states. This is due to the quasiprobability disribution in phasespace of fock states being Gaussian in shape. Another Gaussian state is the coherent state. Fock states have a well defined number of particles in the QHO, while coherent states more closely resemble classic harmonic oscillator behavior.
$$
|\alpha\rangle =  e^{-|\alpha|^2/2}\sum_n \cfrac{\alpha^2}{\sqrt{n!}}|n\rangle \qquad \alpha = \cfrac{\langle \hat{X} \rangle + i \langle \hat{P} \rangle}{\sqrt{2}}
$$
where $\alpha \in \mathbb{C}$. The expected value of an operator can be calculated with:
$$
\Delta A = \sqrt{\langle A^2 \rangle - \langle A \rangle^2}
$$
where
$$
\langle A \rangle = \langle \psi |A|\psi \rangle
$$
The coherent state can also be visualized in the phasespace plot.
